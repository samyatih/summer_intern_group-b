```python
def anomaly_to_probabilities(anomaly):

    # Probability of El Niño
    p_el = expit((anomaly - 0.5) / 0.20)

    # Probability of La Niña
    p_la = expit((-0.5 - anomaly) / 0.20)

    # Remaining probability goes to Neutral
    p_neutral = max(0.0, 1 - p_el - p_la)

    total = p_el + p_neutral + p_la

    p_el /= total
    p_neutral /= total
    p_la /= total

    confidence = max(p_el, p_neutral, p_la)

    return pd.Series({
        "enso_el_nino_probability": p_el,
        "enso_neutral_probability": p_neutral,
        "enso_la_nina_probability": p_la,
        "enso_forecast_confidence": confidence
    })
```
