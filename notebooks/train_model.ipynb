```python
import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import (
    mean_absolute_error,
    root_mean_squared_error
)

import joblib
```

```python
ROOT = Path("..")

PROCESSED_DATA = ROOT / "data" / "processed"
features = pd.read_csv(
    PROCESSED_DATA / "enso_features_multihorizon.csv",
    parse_dates=["Date"]
)
```

```python
feature_cols = [

    "nino34",
    "nino3",
    "nino4",
    "iod",
    "soi",

    "nino34_lag1",
    "nino34_lag2",
    "nino34_lag3",

    "nino3_lag1",
    "nino3_lag2",
    "nino3_lag3",

    "nino4_lag1",
    "nino4_lag2",
    "nino4_lag3",

    "iod_lag1",
    "iod_lag2",
    "iod_lag3",

    "soi_lag1",
    "soi_lag2",
    "soi_lag3",

    "nino34_roll3",
    "nino3_roll3",
    "nino4_roll3",
    "iod_roll3",
    "soi_roll3"
]
```

```python
targets = [

    "future_nino34_lead1",
    "future_nino34_lead2",
    "future_nino34_lead3",
    "future_nino34_lead4",
    "future_nino34_lead5",
    "future_nino34_lead6"

]
```

```python
models = {}

for lead, target in enumerate(targets, start=1):

    X = features[feature_cols]

    y = features[target]

    model = LinearRegression()

    model.fit(X, y)

    models[f"lead{lead}"] = model

    print(f"Lead {lead} model trained.")
```

```python
MODEL_DIR = ROOT / "models"

MODEL_DIR.mkdir(exist_ok=True)

for lead in range(1, 7):

    joblib.dump(

        models[f"lead{lead}"],

        MODEL_DIR /
        f"linear_enso_model_lead{lead}.pkl"

    )

print("All models saved.")
```

```python
joblib.dump(

    feature_cols,

    MODEL_DIR /
    "feature_columns.pkl"

)

print("Feature columns saved.")
```
