```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print(pd.__version__)
```

```python
from pathlib import Path

ROOT = Path("..")

RAW_DATA = ROOT / "data" / "raw"

PROCESSED_DATA = ROOT / "data" / "processed"

print(RAW_DATA)
```

```python
print("Files in raw data folder:\n")

for file in RAW_DATA.iterdir():
    print(file.name)
```

```python
DATA_FILES = {
    "nino34": "nino34.csv",   
    "nino3": "nino3.csv",
    "nino4": "nino4.csv",
    "iod": "iod.csv",
}
```

```python
datasets = {}

for name, filename in DATA_FILES.items():

    file_path = RAW_DATA / filename

    datasets[name] = pd.read_csv(file_path)

print("Datasets Loaded Successfully!")

print(datasets.keys())
```

```python
def inspect_dataset(name, df):

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print(f"Shape : {df.shape}")

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nFirst Five Rows:")
    display(df.head())

    print("\n")

for name, df in datasets.items():

    inspect_dataset(name, df)
```

```python
def inspect(name, df):

    print("="*60)
    print(name.upper())
    print("="*60)

    print(df.head())

    print()

    print(df.info())

    print()

    print(df.describe(include="all"))

for name, df in datasets.items():
    inspect(name, df)
```

```python
def clean_csv_dataset(df, column_name):

    df = df.copy()

    df.columns = ["Date", column_name]

    df["Date"] = pd.to_datetime(df["Date"])

    df[column_name] = pd.to_numeric(df[column_name], errors="coerce")

    df[column_name] = df[column_name].replace(
        [-9999, -9999.0, -99.99],
        np.nan
    )

    df = df.sort_values("Date")

    df.reset_index(drop=True, inplace=True)

    return df
datasets["nino34"] = clean_csv_dataset(
    datasets["nino34"],
    "nino34"
)

datasets["nino3"] = clean_csv_dataset(
    datasets["nino3"],
    "nino3"
)

datasets["nino4"] = clean_csv_dataset(
    datasets["nino4"],
    "nino4"
)

datasets["iod"] = clean_csv_dataset(
    datasets["iod"],
    "iod"
)
datasets["nino34"].head()
```

```python
soi_file = RAW_DATA / "soi.txt"

with open(soi_file, "r") as f:
    lines = f.readlines()

print(len(lines))

```

```python
#parsing soi data
months = [
    "JAN", "FEB", "MAR", "APR", "MAY", "JUN",
    "JUL", "AUG", "SEP", "OCT", "NOV", "DEC"
]

# Find where the standardized table begins
start_idx = None

for i, line in enumerate(lines):
    if "STANDARDIZED" in line.upper():
        start_idx = i
        break

if start_idx is None:
    raise ValueError("STANDARDIZED DATA section not found.")

# Find the YEAR header after STANDARDIZED
year_header = None

for i in range(start_idx, len(lines)):
    if lines[i].strip().startswith("YEAR"):
        year_header = i
        break

if year_header is None:
    raise ValueError("YEAR header not found.")

records = []

# Actual data starts after YEAR header
for line in lines[year_header + 1:]:

    line = line.strip()

    if not line:
        continue

    parts = line.split()

    # Stop when table ends
    if len(parts) != 13 or not parts[0].isdigit():
        break

    year = int(parts[0])

    for month_num, value in enumerate(parts[1:], start=1):

        value = float(value)

        if value == -999.9:
            value = np.nan

        records.append([
            pd.Timestamp(year=year, month=month_num, day=1),
            value
        ])

soi = pd.DataFrame(records, columns=["Date", "soi"])
print(soi.shape)

display(soi.head())

display(soi.tail())

print(soi.isna().sum())
```

```python
#mergin all datasets
master = datasets["nino34"].copy()

for dataset in ["nino3", "nino4", "iod"]:
    master = master.merge(
        datasets[dataset],
        on="Date",
        how="outer"
    )

master = master.merge(
    soi,
    on="Date",
    how="outer"
)

master = master.sort_values("Date").reset_index(drop=True)
master.info()

display(master.head())

display(master.tail())

master.describe()
```

```python
#saving 
master.to_csv(
    PROCESSED_DATA / "master_enso_monthly.csv",
    index=False
)
```
